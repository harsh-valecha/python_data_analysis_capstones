```python
# reading the file using csv 
f = open('student_marks.csv')
read = csv.reader(f)
names = []
for row in read:
    # print(row)
    names.append([row[1],row[-1]])
f.close()
print(names.pop(0))
```

```python
# Sorting by marks
names.sort(key=lambda student:student[1],reverse=True)
print(names[:10])
```

```python
# create a dictionary of students and sort it 
students = {}
with open('student_marks.csv','r') as f:
    rows = csv.reader(f)
    for row in rows:
        students[row[1]]=row[3]
students_lst = list(students.items())[1:]
print(students_lst)  
```

```python
student_sorted = sorted(students_lst,key=lambda item:item[1],reverse=True)[:10]
student_sorted
```

```python
students_lst[:10]
```

```python
# store the sorted values in a csv file
with open('students_sorted_csv.csv','w') as f:
    writer = csv.writer(f,delimiter=',')
    writer.writerows(student_sorted)
```
