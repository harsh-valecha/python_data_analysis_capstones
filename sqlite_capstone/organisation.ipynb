```python
import sqlite3
import pandas as pd
connection  = sqlite3.connect('organisation.db')
cursor = connection.cursor()
```

```python
# getting attendance. > 7 hours
rows = cursor.execute('select * from attendance where hours_worked>7')
for row in rows:
    print(row)
```

```python
# find all the employees whose attendance >6
rows = cursor.execute(''' 
select e.first_name, e.last_name , e.dept_id , e.job_title , e.salary , a.hours_worked
from attendance a inner join
employees e on a.emp_id = e.emp_id
where a.hours_worked > 6
''')
for row in rows:
    print(row)
```

```python
# showing the data in dataframe format
df = pd.read_sql(sql='''
select e.first_name, e.last_name , e.dept_id , e.job_title , e.salary , a.hours_worked
from attendance a inner join
employees e on a.emp_id = e.emp_id
where a.hours_worked > 6
''' , con=connection)
df
```

```python
# select the department where budget is > 3lac
dept_high = pd.read_sql(''' 
select * from departments where 
budget > 350000
''' , connection)
dept_high
```

```python
# show the projects as per department 
pd.read_sql(''' 
select 
d.dept_id as dept_id, d.dept_name as dept_name,
p.project_name as project_name 
, p.budget as project_budget , d.budget as dept_budget  
from departments d
inner join 
projects p 
on 
p.dept_id = d.dept_id
''' , connection)
```

```python
# show all the projects by aggregating over dept_name and show the total costs consumed
pd.read_sql(''' 
select 
d.dept_name , sum(p.budget) as budget_consumed , d.budget as budget_provided
from departments d
inner join 
projects p
on p.dept_id = d.dept_id
group by d.dept_id
''' , connection)
```

```python
# sort the projects by budget desc
pd.read_sql(''' 
select *
from projects
order by budget desc
''' , connection)
```

```python
# find all the high priority tasks of project with project_id=3
high_priority_tasks = pd.read_sql(''' 
select 
t.task_id , p.project_id , p.project_name , t.task_name , t.priority
from tasks t inner join projects p
on t.project_id = p.project_id
where t.priority = 'High'
''' , connection)
high_priority_tasks
```

```python
# export the high priority tasks to a csv file
high_priority_tasks.to_csv('high_priority.csv',index=False)
```

```python
# export high priority tasks to a json file
arr = high_priority_tasks.to_numpy()
arr
```

```python
# preparing a proper dictionary
lst = []
for row in arr:
    value = {}
    value['task_id'] = row[0]
    value['project_id'] = row[1]
    value['project_name'] = row[2]
    value['task_name'] = row[3]
    value['priority'] = row[4]
    lst.append(value)
lst
```

```python
import json
with open('high_priority.json','w') as f:
    json.dump(lst,f,indent=True)
```
