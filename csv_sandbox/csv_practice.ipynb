```python
# printing the columns 
f_opened = open('practice_data.csv','r')
red = csv.reader(f_opened)
lst = list(red)
print(lst[0])
```

```python
# print only 5 rows
five_rows = lst[1:6]
five_rows
```

```python
# calculating average age
ages = list(row[4] for row in lst)[1:]
ages = sum(int(age) for age in ages)
print(ages/len(lst))
```

```python
# counting the people by city 
cities = {}
for row in lst[1:]:
    city = row[5]
    if city in cities:
        cities[city]+=1
    else:
        cities[city]=1
print(len(cities),cities)
```

```python
# finding the maximum and minimum ages
ages_lst = list(int(row[4]) for row in lst[1:])
max_age = max(ages_lst)
min_age = min(ages_lst)
print(max_age,min_age)
```

```python
# finding names of the oldest and youngest people
for row in lst[1:]:
    if row[4]==str(max_age):
        print(f'Max age of :{row[1]}')
    if row[4]==str(min_age):
        print(f'Min age of :{row[1]}')
    
```

```python
# calculating the average salary
avg_salary = 0
for row in lst[1:]:
    avg_salary +=int(row[8])
avg_salary/=len(lst[1:])
print(avg_salary)
```

```python
# count of active , inactive users
active = 0
inactive = 0
for row in lst[1:]:
    if row[-1]=='yes':
        active+=1
    else:
        inactive+=1
print(active,inactive)
```

```python
# highest and lowest paid salaries
salaries = []
for row in lst[1:]:
    salaries.append(int(row[8]))
max_salary = max(salaries)
min_salary = min(salaries)
for row in lst[1:]:
    if int(row[8])==max_salary:
        print(f'Max salary of:{row[2]}')
    if int(row[8])==min_salary:
        print(f'Min salary of:{row[2]}')
```

```python
# filtering the rows for age > 30
age_over_30 = []
for row in lst[1:]:
    if int(row[4])>30:
        age_over_30.append(row)
print(age_over_30)
```

```python
# filtering the country = USA
cntry_USA = []
for row in lst[1:]:
    if row[6]=='USA':
        cntry_USA.append(row)
print(cntry_USA)
```

```python
# reading the csv file and sorting by age desc 
import pandas as pd
df = pd.read_csv('practice_data.csv')
df.head()
```

```python
# sort by age descending 
df.sort_values(by='age',ascending=False)
```

```python
# sort by salary ascending
df.sort_values(by='salary',ascending=False)
```

```python
# writing a new csv with id , first_name , last_name , salary using pandas
df.to_csv('practice_write_pd.csv',columns=['id','first_name','last_name','salary'],index=False)
```

```python
# writing a new csv with id , first_name , last_name , salary
f_writer = open('practice_write_csv.csv','w')
writer = csv.writer(f_writer,delimiter=',')
writer.writerow(['id','first_name','last_name','salary'])
for row in lst[1:]:
    id = row[0]
    first_name = row[1]
    last_name = row[2]
    salary = int(row[8])
    writer.writerow([id,first_name,last_name,salary])
print('done!')    
```
