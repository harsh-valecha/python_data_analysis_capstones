```python
# printing the columns 
f_opened = open('practice_data.csv','r')
red = csv.reader(f_opened)
lst = list(red)
print(lst[0])
```

```python
# print only 5 rows
five_rows = lst[1:6]
five_rows
```

```python
# calculating average age
ages = list(row[4] for row in lst)[1:]
ages = sum(int(age) for age in ages)
print(ages/len(lst))
```

```python
# counting the people by city 
cities = {}
for row in lst[1:]:
    city = row[5]
    if city in cities:
        cities[city]+=1
    else:
        cities[city]=1
print(len(cities),cities)
```

```python
# finding the maximum and minimum ages
ages_lst = list(int(row[4]) for row in lst[1:])
print(max(ages_lst),min(ages_lst))
```
