## Excel Practice

```python
import pandas as pd
```

```python
df = pd.read_excel('product_catalog.xlsx')
df
```

```python
# fetching the rows where product name = Product A
df.loc[
    df['name']=='Product A'
]
```

```python
# fetch the cell value at B2
df.iloc[0, 1]
```

```python
# show all the rows with category= Category A
category_3 = df.loc[
    df['category']=='Category 3'
]
category_3
```

```python
# fetch the sum of category_3
sum(category_3['unitPrice'])
```

```python
# get all the products for category 1 for price less than 100
df.loc[
    df['category']=='Category 1'
].loc[
    df['unitPrice']<100
]
```

```python
# find all the products and there price where price > 100
expensive_products = df.loc[
    df['unitPrice']>100,
    ['name','unitPrice']
]
expensive_products
```

```python
# fetch the row whose product id ends with '010'  
df.loc[
    df['id'].str.endswith('010')
]
```

```python
# export the expensive products to excel
expensive_products.to_excel('expensive_prod.xlsx',index=False)
```
