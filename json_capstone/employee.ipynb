## Employee data JSON Practice

```python
import json
import pandas as pd
pd.options.display.max_rows=50
pd.options.display.max_columns=10
```

```python
with open('employees.json','r') as f:
    data = json.load(f)
data
```

```python
df = pd.read_json('employees.json')
df.head()
```

```python
# sort the employees by salary descending
df.sort_values(by='salary',ascending=False).head()
```

```python
# fetch the employee name , employee_id , hire_date , salary whose salary is greater than 80000
val = []
for e in data:
    if e['salary']>80000:
        val.append((e['employee_id'],e['first_name'],e['last_name'],e['salary'],e['hire_date']))
print(val)
```

```python
df.loc[
    df["salary"] > 80000,
    ["employee_id", "first_name", "last_name", "salary", "hire_date"]
].head()
```

```python
# finding the employees whose hire date was after 1st jan 2008
df.loc[
    pd.to_datetime(df["hire_date"], format="%m/%d/%Y") > pd.Timestamp("2008-01-01"),
    ["employee_id", "first_name", "last_name", "salary", "hire_date"]
].head()
```

```python
df.loc[
    df["department"] == "Finance" ,
    ["employee_id", "first_name", "last_name", "salary", "hire_date","department"]
].loc[
    df["salary"]>10000
]
```

```python
# find all the emails that contains @cbsnews.com
df.loc[df["email"].str.contains("@cbsnews.com", regex=False), ["email"]]
```

```python
finance_people = []
for d in data:
    if d['department']=='Finance' and d['salary']>10000:
        finance_people.append(d)
finance_people
```

```python
with open('finance_json.json','w') as f:
    json.dump(finance_people,f,indent=4)
```
