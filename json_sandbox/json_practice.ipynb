## JSON Practice

```python
import json
from pprint import pprint
```

```python
# loading the json data and priting the company name
f_reader = open('practice_data.json','r')
j_data = json.load(f_reader)
print(f'company name : {j_data['company']['name']}')
```

```python
# counting the total number of employees
print(f'Total Employees:{len(j_data['employees'])}')
```

```python
# finding all the active employees
employees = j_data['employees']
count = 0
for employee in employees:
    if employee['active']==True:
        count+=1
        print(f'employee name:{employee['name']}')
print(f'Total active employees:{count}')
```

```python
# listing the name of employees working in datascience department
for employee in employees:
    if employee['department']=='Data Science':
        print(f'Employee in DataScience:{employee['name']}')
```

```python
# finding the employee with highest rating 
rating = 0
for employee in employees:
    emp_rating = employee['performance']['rating']
    if emp_rating > rating:
        rating = emp_rating
        name = employee['name']
print(f'Highest Rating : {employee['name']} , Rating: {rating}')
```

```python
# printing all the project names
for employee in employees:
    projects = employee['projects']
    for project in projects:
        print(project['project_name'])
```

```python
# counting the tasks marked completed
task_completed = 0
for employee in employees:
    for project in employee['projects']:
        for task in project['tasks']:
            if task['completed']==True:
                task_completed+=1
print(f'Completed tasks:{task_completed}')
```

```python
# extracting skills in a single list
skills = []
for employee in employees:
    skills.extend(employee['skills'])
print(skills)
```

```python
# finding the employee whose remote_ok is true
remote_employees = []
for employee in employees:
    if employee['preferences']['remote_ok']==True:
        remote_employees.append(employee['name'])
print(remote_employees)
```

```python
# identifying the employee with most projects
prj_cnt = 0
for employee in employees:
    projects = employee['projects']
    if len(projects)>prj_cnt:
        prj_cnt = len(projects)
        dedicated = employee['name']
print(f'Employee with most projects:{dedicated} , count:{prj_cnt}')
```

```python
# calculating the average performance rating of all employees
highest_rate = 0
highly_rated = None
for employee in employees:
    emp_rating = employee['performance']['rating']
    if emp_rating > highest_rate:
        highest_rate = emp_rating
        highly_rated = employee['name']
print(f'Highest Rated employee:{highly_rated}, rating:{highest_rate}')
```

```python
# updating one employee's city and saving to disk

for employee in employees:
    if employee['name']=='Charlie Brown':
        employee['city'] = 'Bhatinda'
pprint(employees)
f_write = open('practice_write.json','w')
json.dump(employees,f_write,indent=True)
print('done!')
```
